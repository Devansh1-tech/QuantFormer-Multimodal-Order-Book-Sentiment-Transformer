```python
possible_labels = []

for col in train_df.columns:
    unique_values = sorted(train_df[col].dropna().unique())

    if len(unique_values) <= 5:
        possible_labels.append({
            "Column": col,
            "Unique Values": unique_values
        })

pd.DataFrame(possible_labels)
```
