## Skewness Analysis

The dataset contains both positively and negatively skewed features.

A few features have very high positive skewness, indicating rare but large market events.

This behavior is common in financial order-book data.

Since the dataset is already normalized and deep learning models do not require normally distributed inputs, no skewness correction will be applied.

```python
sample_features = train_df.columns[:10]

plt.figure(figsize=(16,8))

train_df[sample_features].boxplot(rot=90)

plt.title("Boxplots of First 10 Features")
plt.show()
```

```python
outlier_summary = {}

for col in train_df.columns[:20]:

    Q1 = train_df[col].quantile(0.25)
    Q3 = train_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = train_df[
        (train_df[col] < lower) |
        (train_df[col] > upper)
    ].shape[0]

    outlier_summary[col] = outliers

pd.DataFrame(
    outlier_summary.items(),
    columns=["Feature","Outlier Count"]
)
```

## Outlier Analysis

Some features contain outliers.

This is expected in financial market data.

These values represent real market movements.

We will not remove them because they contain useful information.

```python
sample_features = train_df.columns[:10]

plt.figure(figsize=(15, 6))

train_df[sample_features].boxplot(rot=90)

plt.title("Boxplot of First 10 Features")
plt.ylabel("Feature Value")
plt.show()
```

```python
outlier_summary = []

for col in train_df.columns[:20]:

    Q1 = train_df[col].quantile(0.25)
    Q3 = train_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((train_df[col] < lower) | (train_df[col] > upper)).sum()

    outlier_summary.append({
        "Feature": col,
        "Outlier Count": outliers,
        "Outlier %": round((outliers / len(train_df)) * 100, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df
```

## Outlier Analysis

Outliers are present in several features.

This is expected in financial market data because large buy and sell orders occur naturally.

These outliers represent genuine market behavior rather than data errors.

Therefore, outliers will not be removed during preprocessing.

```python
# Check duplicate columns

duplicate_columns = []

cols = train_df.columns[:143]

for i in range(len(cols)):
    for j in range(i + 1, len(cols)):
        if train_df[cols[i]].equals(train_df[cols[j]]):
            duplicate_columns.append((cols[i], cols[j]))

print("Duplicate Columns:", duplicate_columns)
```

## Duplicate Feature Analysis

Some input features are exact duplicates of each other.

This duplication is part of the original FI-2010 dataset.

To preserve the benchmark dataset and ensure reproducibility, these features will be retained during model training.

No duplicate feature removal will be performed in the baseline implementation.
